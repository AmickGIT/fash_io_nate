```python
import csv
import re

input_file = "data.csv"
output_file = "new_data.csv"

with open(input_file, "r", newline='') as infile, open(output_file, "w", newline='') as outfile:
    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    for row in reader:
        cleaned_row = []
        for cell in row:
            # Step 1: Strip leading/trailing whitespace first
            cell = cell.strip()
            # Step 2: Remove square brackets
            cell = re.sub(r'[\[\]]', '', cell)
            # Step 3: Remove whitespace after commas
            cell = re.sub(r',\s+', ',', cell)
            cleaned_row.append(cell)
        writer.writerow(cleaned_row)
```

```python
import csv

# Input and output file paths
input_file = 'new_data.csv'
output_file = 'cleaned_data.csv'

with open(input_file, mode='r', newline='', encoding='utf-8') as infile, \
     open(output_file, mode='w', newline='', encoding='utf-8') as outfile:
    
    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    for row in reader:
        lower_row = [cell.lower() for cell in row]
        writer.writerow(lower_row)
```

```python
import pandas as pd

# Load the main and exclusion CSV files
main_df = pd.read_csv("new_data.csv")
exclude_df = pd.read_csv("description.csv")

# Get the set of 'img' values to exclude
exclude_imgs = set(exclude_df['img'])

# Filter the main DataFrame to exclude rows with those 'img' values
filtered_df = main_df[~main_df['img'].isin(exclude_imgs)]

# Save the result to a new CSV file (optional)
filtered_df.to_csv("filtered_desc.csv", index=False)

print("Rows removed:", len(main_df) - len(filtered_df))
```

```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("new_data.csv")

# Drop duplicates based on the 'img' column, keeping the first occurrence
df_clean = df.drop_duplicates(subset='img', keep='first')

# Save the cleaned CSV (optional)
df_clean.to_csv("newer_file.csv", index=False)

print(f"Removed {len(df) - len(df_clean)} duplicate rows based on 'img'.")
```
