```python
import csv
import re

input_file = "data.csv"
output_file = "new_data.csv"

with open(input_file, "r", newline='') as infile, open(output_file, "w", newline='') as outfile:
    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    for row in reader:
        cleaned_row = []
        for cell in row:
            # Step 1: Strip leading/trailing whitespace first
            cell = cell.strip()
            # Step 2: Remove square brackets
            cell = re.sub(r'[\[\]]', '', cell)
            # Step 3: Remove whitespace after commas
            cell = re.sub(r',\s+', ',', cell)
            cleaned_row.append(cell)
        writer.writerow(cleaned_row)
```

```python
import csv

# Input and output file paths
input_file = 'new_data.csv'
output_file = 'cleaned_data.csv'

with open(input_file, mode='r', newline='', encoding='utf-8') as infile, \
     open(output_file, mode='w', newline='', encoding='utf-8') as outfile:
    
    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    for row in reader:
        lower_row = [cell.lower() for cell in row]
        writer.writerow(lower_row)
```

```python
import pandas as pd

def quote_if_needed(text):
    if pd.isna(text):
        return text  # Leave NaN values unchanged
    text = str(text).strip()
    if not (text.startswith('"') and text.endswith('"')):
        return f'"{text}"'
    return text

# Load the original CSV
file_path = "description.csv"
df = pd.read_csv(file_path)

# Apply the quote-fix function
df['description'] = df['description'].apply(quote_if_needed)

# Overwrite the original CSV
df.to_csv(file_path, index=False)

print("Updated 'description' column in-place with double quotes where needed.")
```

```python
import pandas as pd
import csv

# Load CSV
df = pd.read_csv("description.csv", dtype=str)

# Optional: Clean up quotes inside, but do NOT wrap the whole string
def fix_inner_quotes(text):
    if pd.isna(text):
        return text
    return str(text).replace('"', '""')  # double quotes inside

df['description'] = df['description'].apply(fix_inner_quotes)

# Write the CSV with proper quoting
df.to_csv("quoted_desc.csv", index=False, quoting=csv.QUOTE_ALL)
```

```python
import pandas as pd
import csv

def fix_description(text):
    if pd.isna(text):
        return text
    text = str(text).strip()

    # Remove outer quotes if they exist
    if text.startswith('"') and text.endswith('"'):
        text = text[1:-1]

    # Escape inner quotes
    text = text.replace('"', '""')

    # Wrap in a single set of double quotes
    return f'"{text}"'

# Load CSV as strings to preserve formatting
df = pd.read_csv("description.csv", dtype=str)

# Apply only to 'description' column
df['description'] = df['description'].apply(fix_description)

# Overwrite CSV (no auto-quoting; 'description' already has correct quotes)
df.to_csv("quoted_desc_new.csv", index=False, quoting=csv.QUOTE_NONE, escapechar='\\')
```
